```python
# Microatividade 1: Ler um arquivo CSV usando Pandas

import pandas as pd
import io

# Lendo o arquivo físico, removendo o BOM, aspas duplas externas e aspas simples das datas
with open('dados.csv', 'r', encoding='utf-8-sig') as f:
    linhas_limpas = []
    for linha in f:
        linha = linha.strip()
        # Remove as aspas duplas que envelopam a linha inteira se existirem
        if (linha.startswith('"') and linha.endswith('"')) or (linha.startswith("'") and linha.endswith("'")):
            linha = linha[1:-1]
        # Remove aspas simples internas (como as que envolvem as datas)
        linha = linha.replace("'", "")
        linhas_limpas.append(linha)

# Cria o DataFrame a partir das linhas limpas usando o delimitador ';'
conteudo_limpo = "\n".join(linhas_limpas)
df = pd.read_csv(io.StringIO(conteudo_limpo), sep=';')

# Exibindo os dados corrigidos
print(df)
```

```python
# Microatividade 2: Criar um subconjunto de dados

# Criando uma nova variável com apenas 3 colunas do DataFrame original
df_subset = df[['Duration', 'Pulse', 'Calories']]

# Exibindo o subconjunto
print(df_subset)
```

```python
import pandas as pd
import io

# 1 a 4. Leitura do arquivo CSV de forma robusta e limpa para evitar o erro de substring do motor python
with open('dados.csv', 'r', encoding='utf-8-sig') as f:
    linhas_limpas = []
    for linha in f:
        linha = linha.strip()
        if (linha.startswith('"') and linha.endswith('"')) or (linha.startswith("'") and linha.endswith("'")):
            linha = linha[1:-1]
        linha = linha.replace("'", "")
        linhas_limpas.append(linha)

conteudo_limpo = "\n".join(linhas_limpas)
df = pd.read_csv(io.StringIO(conteudo_limpo), sep=';')

# 5. Verificando se os dados foram importados adequadamente
print("--- Informações Gerais do Dataset Original ---")
df.info()

print("\n--- Primeiras e Últimas Linhas ---")
print(df.head(5))
print(df.tail(5))

# 6 a 10. Criando uma cópia do conjunto de dados original
df_limpo = df.copy()

# 12. Substituindo valores nulos da coluna 'Calories' por 0
df_limpo['Calories'] = df_limpo['Calories'].fillna(0)
print("\n--- DataFrame após preencher 'Calories' com 0 ---")
print(df_limpo)

# 13. Substituindo valores nulos da coluna 'Date' por '1900/01/01'
df_limpo['Date'] = df_limpo['Date'].fillna('1900/01/01')
print("\n--- DataFrame após preencher 'Date' com '1900/01/01' ---")
print(df_limpo)

# Tentativa de transformar 'Date' em datetime (gerará erro devido ao formato de '1900/01/01' e '20201226')
# df_limpo['Date'] = pd.to_datetime(df_limpo['Date'], format='%Y/%m/%d')

# 14. Substituindo '1900/01/01' por NaN na coluna 'Date'
df_limpo['Date'] = df_limpo['Date'].replace('1900/01/01', pd.NA)

# 20. Tratando o valor sem barras "20201226" combinando replace e to_datetime
df_limpo['Date'] = df_limpo['Date'].replace('20201226', '2020/12/26')

# 21 a 22. Transformando toda a coluna 'Date' para o formato datetime
df_limpo['Date'] = pd.to_datetime(df_limpo['Date'], format='%Y/%m/%d')

print("\n--- DataFrame após conversão da coluna 'Date' para datetime ---")
print(df_limpo)

# 24 a 26. Removendo os registros contendo valores nulos (linha 22 na coluna 'Date')
df_limpo.dropna(subset=['Date'], inplace=True)

# 27. Imprimindo o DataFrame final limpo e transformado
print("\n--- DataFrame Final Tratado e Limpo ---")
print(df_limpo)
df_limpo.info()
```
